```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# FEATURE AUDIT
# ---------------------------------------------------------

print("=" * 80)
print("FEATURE AUDIT")
print("=" * 80)

# X_train is already the transformed data.
# Get feature names from the preprocessor if available.

try:
    feature_names = data_transformer.preprocessor.get_feature_names_out()
except Exception:
    feature_names = [
        f"Feature_{i}"
        for i in range(X_train.shape[1])
    ]

print("\nNumber of model features:", len(feature_names))

# ---------------------------------------------------------
# 1. CONSTANT FEATURES
# ---------------------------------------------------------

X_train_df = pd.DataFrame(
    X_train,
    columns=feature_names
)

constant_features = [
    col
    for col in X_train_df.columns
    if X_train_df[col].nunique() <= 1
]

print("\n" + "-" * 80)
print("1. CONSTANT FEATURES")
print("-" * 80)

if constant_features:
    print(constant_features)
else:
    print("No constant features found.")


# ---------------------------------------------------------
# 2. MISSING VALUES AFTER TRANSFORMATION
# ---------------------------------------------------------

print("\n" + "-" * 80)
print("2. MISSING VALUES")
print("-" * 80)

missing_count = X_train_df.isnull().sum().sum()

print("Total missing values:", missing_count)

if missing_count == 0:
    print("No missing values after preprocessing.")


# ---------------------------------------------------------
# 3. INFINITE VALUES
# ---------------------------------------------------------

print("\n" + "-" * 80)
print("3. INFINITE VALUES")
print("-" * 80)

infinite_count = np.isinf(
    X_train_df.to_numpy()
).sum()

print("Total infinite values:", infinite_count)


# ---------------------------------------------------------
# 4. HIGHLY CORRELATED FEATURES
# ---------------------------------------------------------

print("\n" + "-" * 80)
print("4. HIGHLY CORRELATED FEATURES")
print("-" * 80)

correlation_matrix = X_train_df.corr().abs()

upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

high_correlation_pairs = []

for column in upper_triangle.columns:

    for row in upper_triangle.index:

        value = upper_triangle.loc[row, column]

        if pd.notna(value) and value >= 0.90:

            high_correlation_pairs.append(
                (
                    row,
                    column,
                    round(value, 4)
                )
            )

high_correlation_pairs = sorted(
    high_correlation_pairs,
    key=lambda x: x[2],
    reverse=True
)

if high_correlation_pairs:

    for pair in high_correlation_pairs[:30]:
        print(pair)

else:
    print("No feature pairs with correlation >= 0.90.")


# ---------------------------------------------------------
# 5. FEATURE VARIANCE
# ---------------------------------------------------------

print("\n" + "-" * 80)
print("5. LOW VARIANCE FEATURES")
print("-" * 80)

variance = X_train_df.var().sort_values()

print(
    variance.head(15)
)


# ---------------------------------------------------------
# 6. FEATURE SUMMARY
# ---------------------------------------------------------

print("\n" + "-" * 80)
print("6. FEATURE SUMMARY")
print("-" * 80)

summary = pd.DataFrame({
    "Feature": X_train_df.columns,
    "Mean": X_train_df.mean().values,
    "Std": X_train_df.std().values,
    "Min": X_train_df.min().values,
    "Max": X_train_df.max().values,
    "Unique": X_train_df.nunique().values
})

print(
    summary.head(20).to_string(index=False)
)

print("\nFeature audit completed.")
```

```python
# ============================================================
# FRAUD vs NON-FRAUD FEATURE ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

df = provider_features.copy()

# Convert target to numeric temporarily
df["Fraud"] = (
    df["PotentialFraud"]
    .astype(str)
    .str.lower()
    .map({
        "yes": 1,
        "no": 0
    })
)

# Top features from model importance
top_features = (
    importance_df
    .head(20)["Feature"]
    .tolist()
)

print("=" * 100)
print("FRAUD vs NON-FRAUD FEATURE ANALYSIS")
print("=" * 100)

comparison = []

for feature in top_features:

    fraud_values = df.loc[
        df["Fraud"] == 1,
        feature
    ]

    nonfraud_values = df.loc[
        df["Fraud"] == 0,
        feature
    ]

    comparison.append({
        "Feature": feature,

        "Fraud_Median":
            fraud_values.median(),

        "NonFraud_Median":
            nonfraud_values.median(),

        "Fraud_Mean":
            fraud_values.mean(),

        "NonFraud_Mean":
            nonfraud_values.mean(),

        "Missing_Fraud_%":
            fraud_values.isna().mean() * 100,

        "Missing_NonFraud_%":
            nonfraud_values.isna().mean() * 100
    })

comparison_df = pd.DataFrame(comparison)

print(
    comparison_df.to_string(
        index=False
    )
)
```

```python
comparison_df["Mean_Difference_%"] = (
    (
        comparison_df["Fraud_Mean"]
        -
        comparison_df["NonFraud_Mean"]
    )
    /
    comparison_df["NonFraud_Mean"].abs().replace(
        0,
        np.nan
    )
) * 100

print(
    comparison_df[
        [
            "Feature",
            "Fraud_Mean",
            "NonFraud_Mean",
            "Mean_Difference_%"
        ]
    ]
    .sort_values(
        "Mean_Difference_%",
        ascending=False
    )
    .to_string(index=False)
)
```

```python
import numpy as np
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Get test probabilities from the selected model
test_prob = best_model.predict_proba(X_test)[:, 1]

threshold_results = []

for threshold in np.arange(0.05, 0.96, 0.01):

    predictions = (
        test_prob >= threshold
    ).astype(int)

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    })

threshold_df = pd.DataFrame(
    threshold_results
)

print(
    threshold_df
    .sort_values("F1", ascending=False)
    .head(15)
    .to_string(index=False)
)
```

```python
print("Maximum F1:", threshold_df["F1"].max())

print("\nBest precision:")
print(
    threshold_df
    .sort_values("Precision", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\nThresholds with Precision >= 0.80:")
print(
    threshold_df[
        threshold_df["Precision"] >= 0.80
    ]
    .sort_values("F1", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\nThresholds with Recall >= 0.90:")
print(
    threshold_df[
        threshold_df["Recall"] >= 0.90
    ]
    .sort_values("F1", ascending=False)
    .head(10)
    .to_string(index=False)
)
```
